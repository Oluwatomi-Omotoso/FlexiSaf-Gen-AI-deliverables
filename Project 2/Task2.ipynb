```python
import pandas as pd
import matplotlib.pyplot as plt

# Load dataset directly
data = "./company_sales_data.csv"
data = pd.read_csv(data)

months = data['month_number']
profit = data['total_profit']

plt.plot(months, profit, marker='o', color='blue')
plt.title("Total Profit per Month")
plt.xlabel("Month")
plt.ylabel("Profit ($)")
plt.xticks(months)
plt.grid(True)

plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

data_url = "./company_sales_data.csv"
data = pd.read_csv(data_url)

months = data['month_number']
soap = data['bathingsoap']
facewash = data['facewash']

# Subplot 1: Bathing Soap
plt.subplot(1, 2, 1)
plt.plot(months, soap, marker='o', color='darkgreen')
plt.title("Bathing Soap Sales")
plt.xlabel("Month")
plt.ylabel("Units")
plt.xticks(months)
plt.grid(True)

# Subplot 2: Facewash
plt.subplot(1, 2, 2)
plt.plot(months, facewash, marker='o', color='orange')
plt.title("Facewash Sales")
plt.xlabel("Month")
plt.ylabel("Units")
plt.xticks(months)
plt.grid(True)

plt.tight_layout()
plt.show()
```
