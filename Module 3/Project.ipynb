```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
```

1. Create a basic line plot

```python
# Only one line
a = np.array([1,2,3,4,5])
plt.plot(a)
plt.show()


# Only one line
b = np.array([2,2,2,2,2])
plt.plot(b)
plt.show()
```

```python
# Both plots sharing a line
plt.figure(figsize = (14,9))
plt.plot(a, color = 'green', linestyle = '-.', label = 'plot_1')
plt.plot(b, linewidth = 5, label = 'plot_2')
plt.title('Merged plots', fontsize = 20, fontweight = 'bold')
plt.legend(loc = 'upper center')
plt.show()
```

Visualising a dataset using Seaborn

```python
import seaborn as sns
# Dataset for this visualisation would be a prepackaged dataset that comes with seaborn.
df = sns.load_dataset('tips')
df.head()
```

```python
# Histogram Plot using seaborn
sns.histplot(data = df, x= 'tip', bins=10, color = "green", )
```

```python
# Histogram plot including a kernel Density estimate to better visualise the shape of the distribution
sns.histplot(data=df, x = 'tip', bins = 10, kde = True, color ='green')
plt.title('Distribution of Total Tips')
plt.xlabel('Total Tips ($)')
plt.ylabel('Frequency')
plt.show()
```

```python
# Creating a Bar plot using seaborn showing the average total_bill for each day.

sns.barplot(data=df, x='day', y='total_bill', errorbar='sd', palette='viridis')

plt.title('Average Total Bill by Day')
plt.xlabel('Day of the Week')
plt.ylabel('Average Total Bill ($)')
plt.show()
```
