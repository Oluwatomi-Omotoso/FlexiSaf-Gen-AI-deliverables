```python
import pandas as pd
import numpy as np
```

Data Preparation

```python
data_source = pd.read_csv('./Source/adult11.csv')

data_source.head()
```

Exploring the dataset

```python
col = [c for c in data_source.columns]
print (f"Columns found inside the dataset are: {col}")
```

```python
print(data_source.info())
```

```python
print(data_source.describe())
```

```python
print(data_source.dtypes)
```

```python
integer_types = [col for col, dtype in data_source.dtypes.items() if np.issubdtype(dtype, np.integer)]
print(f"Numerical columns are: {integer_types}")
```

Statistics calculations on numerical data types

```python
numeric_data_values = data_source.select_dtypes(include = 'number')
mean_values = numeric_data_values.mean()
print(f"MEAN VALUES\n {mean_values}\n")
median_values = numeric_data_values.median()
print(f"MEDIAN VALUES\n{median_values}\n")
mode_values = numeric_data_values.mode()
print(f"MODE VALUES\n{mode_values}\n")
sd_values = numeric_data_values.std()
print(f"STANDARD DEVIATION\n{sd_values}\n")
```

Correlation between features

```python
correlation_matrix = numeric_data_values.corr()
correlation_matrix
```
